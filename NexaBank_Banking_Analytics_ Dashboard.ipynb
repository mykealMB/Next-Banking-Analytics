```python
import pandas as pd
import numpy as np

# Set seed so the results are reproducible
np.random.seed(42)

# Number of accounts
n_accounts = 500

# Create account IDs
account_ids = [f"NB{str(i).zfill(5)}" for i in range(1, n_accounts + 1)]

# Account types
account_types = np.random.choice(["Savings", "Current", "Salary"],
    size=n_accounts,
    p=[0.50, 0.30, 0.20])

# Generate account balances
balances = np.random.randint(5000, 5000000, size=n_accounts)

# Generate savings balances
savings_balances = np.random.randint(0, 2500000, size=n_accounts)

# Generate loan balances
loan_balances = np.random.choice([0, 0, 0, 50000, 100000, 250000, 500000, 750000, 1000000, 1500000],
    size=n_accounts)

# Account status
account_status = np.random.choice(["Active", "Inactive"],
    size=n_accounts,
    p=[0.92, 0.08])

# Generate account opening dates
date_opened = pd.to_datetime(np.random.choice(pd.date_range("2021-01-01", "2026-06-30"),size=n_accounts))

# Create the accounts DataFrame
accounts = pd.DataFrame({
    "Customer_ID": [f"CUST{str(i).zfill(5)}" for i in range(1, n_accounts + 1)],
    "Account_ID": account_ids,
    "Account_Type": account_types,
    "Balance": balances,
    "Savings_Balance": savings_balances,
    "Loan_Balance": loan_balances,
    "Account_Status": account_status,
    "Date_Opened": date_opened})

# Display the first 10 rows
accounts.head(10)
```

```python
# Number of transactions
n_transactions = 5000

# Generate transaction IDs
transaction_ids = [f"TXN{str(i).zfill(6)}" for i in range(1, n_transactions + 1)]

# Randomly assign transactions to existing accounts
transaction_account_ids = np.random.choice(accounts["Account_ID"],size=n_transactions)

# Generate transaction dates
transaction_dates = pd.to_datetime(np.random.choice(pd.date_range("2025-01-01", "2026-06-30"),size=n_transactions))

# Transaction types
transaction_types = np.random.choice(["Credit", "Debit"],size=n_transactions,p=[0.40, 0.60])

# Transaction categories
categories = np.random.choice(
    [
        "Food",
        "Bills",
        "Shopping",
        "Transfer",
        "Salary",
        "ATM Withdrawal",
        "Transport",
        "Healthcare",
        "Entertainment",
        "Savings"
    ],
    size=n_transactions)

# Generate transaction amounts
amounts = np.random.randint(1000,500000,size=n_transactions)

# Payment methods
payment_methods = np.random.choice(
    ["Card", "Bank Transfer", "ATM", "USSD", "Mobile Banking"],size=n_transactions)

# Transaction status
statuses = np.random.choice(["Successful", "Pending", "Failed"],size=n_transactions,p=[0.94, 0.04, 0.02])

# Create the transactions DataFrame
transactions = pd.DataFrame({
    "Transaction_ID": transaction_ids,
    "Account_ID": transaction_account_ids,
    "Transaction_Date": transaction_dates,
    "Transaction_Type": transaction_types,
    "Category": categories,
    "Amount": amounts,
    "Payment_Method": payment_methods,
    "Status": statuses
})

# Display the first 10 transactions
transactions.head(10)
```

## 3.2 Spending by Category

This analysis examines debit transactions by spending category to identify the areas where customers spend the most money.

Understanding spending patterns will help identify the major expense categories and provide useful information for the banking dashboard.

```python
# Spending by Category

debit_transactions = transactions[transactions["Transaction_Type"] == "Debit"]

spending_by_category = (debit_transactions.groupby("Category")["Amount"].sum().sort_values(ascending=False))

spending_by_category
```

```python
spending_percentage = (spending_by_category / spending_by_category.sum() * 100).round(2)
spending_percentage
```

```python
# Correct transaction categories based on transaction type

credit_categories = ["Salary","Transfer","Refund","Interest","Savings Deposit"]

debit_categories = ["Food","Bills","Shopping","Transfer","ATM Withdrawal","Transport","Healthcare","Entertainment"]

# Assign appropriate categories based on transaction type
transactions.loc[transactions["Transaction_Type"] == "Credit","Category"] = np.random.choice(credit_categories,
    size=(transactions["Transaction_Type"] == "Credit").sum())

transactions.loc[transactions["Transaction_Type"] == "Debit","Category"] = np.random.choice(debit_categories,
    size=(transactions["Transaction_Type"] == "Debit").sum())

# Check the updated categories
transactions.groupby(["Transaction_Type", "Category"]).size()
```

## 3.3 Monthly Transaction Trends

This analysis examines transaction activity over time by grouping transactions by month.

The analysis will help identify changes in transaction volume and transaction value across the period covered by the dataset.

```python
# Convert transaction date to monthly period

transactions["Month"] = transactions["Transaction_Date"].dt.to_period("M")

monthly_transactions = (transactions.groupby("Month").agg(Transaction_Count=("Transaction_ID", "count"),
        Transaction_Value=("Amount", "sum")))

monthly_transactions
```

```python
# credits and debits by month:


monthly_flow = (transactions.groupby(["Month", "Transaction_Type"])["Amount"].sum().unstack(fill_value=0))

monthly_flow
```
